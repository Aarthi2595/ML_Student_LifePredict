# STUDENT LIFESTYLE AND ACADEMIC PERFORMANCE PREDICTION

**Machine Learning Project | Random Forest Regression**

This project explores the relationship between students' daily lifestyle habits and academic performance. A Random Forest Regression model is trained to estimate a student's Grade Point Average (GPA) from five lifestyle features.

> **Important:** This model estimates GPA from patterns in the supplied dataset. It does not prove that lifestyle habits cause a particular GPA.


## 1. Introduction

Machine Learning is a branch of Artificial Intelligence that enables computers to learn patterns from data and make predictions. Students' academic performance may be associated with several aspects of daily life, including study time, sleep, extracurricular activities, social activities and physical activity.

This project uses student lifestyle data to build a regression model that estimates GPA from selected daily lifestyle habits.


## 2. Problem Statement

Traditional academic performance evaluation generally focuses on marks, grades and examination results. Lifestyle information may provide additional context about students' routines.

The problem addressed in this project is to develop a Machine Learning model that estimates students' GPA using their daily lifestyle information.


## 3. Objectives

- Analyze student lifestyle data.
- Identify the lifestyle features associated with GPA in the dataset.
- Apply a Machine Learning regression algorithm.
- Predict GPA using five selected lifestyle features.
- Evaluate the model using MAE, RMSE and R² score.


## 4. Existing System

Academic performance is commonly assessed using examination marks, assignments, attendance and grades. These measures describe academic outcomes, but may not combine daily lifestyle information in a single predictive analysis.

**Limitations of the existing approach considered in this project:**
- Lifestyle factors may not be analyzed together.
- It may be difficult to explore patterns across many student records manually.
- A simple marks-based view may not provide an estimate based on daily routine information.

These points describe the motivation for this project; they do not mean that all existing academic systems ignore lifestyle factors.


## 5. Proposed System

The proposed system uses a Random Forest Regression model to estimate GPA from five lifestyle inputs:

1. Study hours per day
2. Extracurricular hours per day
3. Sleep hours per day
4. Social hours per day
5. Physical activity hours per day

The dataset is inspected, the selected features and GPA target are separated, and the records are divided into training and testing sets. The model is trained on the training set and evaluated on unseen test records. A sample student profile is then passed to the trained model to obtain an estimated GPA.

## 9. Implementation

Run the following cells from top to bottom. The CSV-loading cell first checks the notebook folder and then the common Google Colab `/content` folder. If the file is not found, upload `student_lifestyle_dataset.csv` to the notebook session.


```python
# Import required libraries
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load the dataset from the notebook folder or Google Colab's /content folder
possible_paths = [
    Path("student_lifestyle_dataset.csv"),
    Path("/content/student_lifestyle_dataset.csv")
]

csv_file = next((path for path in possible_paths if path.exists()), None)

if csv_file is None:
    raise FileNotFoundError(
        "Dataset not found. Upload student_lifestyle_dataset.csv "
        "to the notebook folder or to /content in Google Colab."
    )

data = pd.read_csv(csv_file)
print("Dataset loaded successfully:", csv_file)
data.head()
```

```python
# Inspect dataset structure and quality
print("Dataset shape:", data.shape)
print("\nColumn names:")
print(data.columns.tolist())

print("\nDataset information:")
data.info()

print("\nMissing values:")
print(data.isnull().sum())

print("\nDuplicate rows:", data.duplicated().sum())
```

```python
# Summary statistics for the dataset
data.describe()
```

```python
# Select the five lifestyle features and GPA target
feature_columns = [
    "Study_Hours_Per_Day",
    "Extracurricular_Hours_Per_Day",
    "Sleep_Hours_Per_Day",
    "Social_Hours_Per_Day",
    "Physical_Activity_Hours_Per_Day"
]

X = data[feature_columns]
y = data["GPA"]

print("Input data shape:", X.shape)
print("Target data shape:", y.shape)
X.head()
```

```python
# Split the dataset: 80% for training and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
# Create and train the Random Forest Regression model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)
print("Model training completed successfully.")
```

```python
# Predict GPA for the test dataset
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual GPA": y_test.values,
    "Predicted GPA": y_pred
})

results.head(10)
```

## 10. Results and Evaluation

The model is evaluated on the test dataset, which was not used during training.

- **MAE (Mean Absolute Error):** Average absolute difference between actual and predicted GPA. Lower is better.
- **RMSE (Root Mean Squared Error):** Penalizes larger prediction errors more strongly. Lower is better.
- **R² score:** Indicates how much variation in the target is explained by the model. A value closer to 1 generally indicates a better fit; it can also be negative.


```python
# Calculate model evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION RESULTS")
print("------------------------")
print(f"MAE      : {mae:.3f}")
print(f"RMSE     : {rmse:.3f}")
print(f"R² Score : {r2:.3f}")
```

### 10.1 GPA Distribution

This chart shows how GPA values are distributed in the dataset.


```python
plt.figure(figsize=(8, 5))
plt.hist(data["GPA"], bins=20, edgecolor="black")
plt.title("Distribution of Student GPA")
plt.xlabel("GPA")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

### 10.2 Actual GPA vs Predicted GPA

Points closer to the diagonal line represent predictions closer to the actual GPA values.


```python
plt.figure(figsize=(7, 6))
plt.scatter(y_test, y_pred, alpha=0.65)

line_min = min(y_test.min(), y_pred.min())
line_max = max(y_test.max(), y_pred.max())
plt.plot([line_min, line_max], [line_min, line_max], linestyle="--")

plt.title("Actual GPA vs Predicted GPA")
plt.xlabel("Actual GPA")
plt.ylabel("Predicted GPA")
plt.grid(alpha=0.3)
plt.show()
```

### 10.3 Feature Importance

Feature importance indicates how much each input feature contributes to the fitted Random Forest model's predictions. It is a model-based measure, not proof of cause and effect.


```python
importance = pd.Series(
    model.feature_importances_,
    index=feature_columns
).sort_values()

plt.figure(figsize=(9, 5))
importance.plot(kind="barh")
plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Lifestyle Feature")
plt.grid(axis="x", alpha=0.3)
plt.show()

print(importance.sort_values(ascending=False))
```

### 10.4 Correlation Heatmap

The correlation matrix shows pairwise linear correlations between numeric columns. Correlation does not establish causation.


```python
numeric_data = data.select_dtypes(include="number")
correlation = numeric_data.corr()

fig, ax = plt.subplots(figsize=(9, 7))
image = ax.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)

ax.set_xticks(range(len(correlation.columns)))
ax.set_yticks(range(len(correlation.columns)))
ax.set_xticklabels(correlation.columns, rotation=45, ha="right")
ax.set_yticklabels(correlation.columns)
ax.set_title("Correlation Heatmap")

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        ax.text(j, i, f"{correlation.iloc[i, j]:.2f}",
                ha="center", va="center", fontsize=8)

fig.colorbar(image, ax=ax, label="Correlation")
plt.tight_layout()
plt.show()
```

### 10.5 Sample Student GPA Prediction

The following example predicts GPA for a sample student. The input values are hours per day and should be between 0 and 24, with their total not exceeding 24 hours.


```python
# Sample student lifestyle information (hours per day)
new_student = pd.DataFrame([{
    "Study_Hours_Per_Day": 4,
    "Extracurricular_Hours_Per_Day": 2,
    "Sleep_Hours_Per_Day": 8,
    "Social_Hours_Per_Day": 3,
    "Physical_Activity_Hours_Per_Day": 1
}], columns=feature_columns)

# Basic validation of the sample input
if (new_student.iloc[0] < 0).any() or (new_student.iloc[0] > 24).any():
    raise ValueError("Each daily activity value must be between 0 and 24 hours.")

if new_student.iloc[0].sum() > 24:
    raise ValueError("The total of daily activity hours cannot exceed 24.")

predicted_gpa = model.predict(new_student)[0]
print("Predicted GPA:", round(predicted_gpa, 2))
```
